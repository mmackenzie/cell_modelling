```python
import pickle
import pandas as pd
import numpy as np
```

```python
# Open pickle file
file_path = r"C:\Users\mmackenzie\Downloads\CALB_0_B182.pkl"

with open(file_path, "rb") as f:
    data_dict = pickle.load(f)

print("✅ Pickle loaded successfully.")
print("Type:", type(data_dict))
print("Keys:", list(data_dict.keys())[:10])  # show first few keys
```

```python
for k, v in data_dict.items():
    try:
        print(f"{k}: length = {len(v)}")
        print(f"Value = {v}")
    except TypeError:
        print(f"{k}: not a list/array (type = {type(v)})")
```

```python
# Convert to DataFrame
if all(isinstance(v, (list, np.ndarray)) for v in data_dict.values()):
    df = pd.DataFrame(data_dict)
else:
    df = pd.DataFrame(data_dict).T

# Show info and first rows
print("\nDataFrame shape:", df.shape)
print(df.head())

```
